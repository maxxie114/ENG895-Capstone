```python
from google.colab import drive
drive.mount('/content/drive')

recs = [
  # ADJECTIVAL RECURSION
  "Jack van der Molen might be the sneaky old thief.",
  "Now the six trainees construct a medium-sized sturdy lamp.",
  "Luna has been chasing the small white mouse all day long.",
  "Unidentified flying objects may have conflicting characteristics.",
  "My cranky geriatric manager is beyond incompetent.",
  "A small child can easily get lost in a vast American mall.",

  # POSSESSIVE RECURSION
  "A sudden shriek penetrated the cold winter's night's dead silence.",
  "The Old Continent's civilizations' advances have shaped history.",
  "This town's only road's potholes are driving me insane.",
  "The footprints clearly revealed Beth's young companion's escape route.",
  "He spotted the infamous manuscript in the wizard's library's dark corner.",
  "Do I dare challenge my advisor's thesis' central argument?",

  # PP RECURSION
  "She saw someone loitering by the entrance to the library.",
  "We sow fear in the hearts of our enemies.",
  "He had been warned about the dangers of tobacco.",
  "There is a little treat for you in the cupboard between the lamps.",
  "Even their recipe for the basic dough of the pie crust is heavily guarded.",
  "Drowning your sorrows in a glass of whiskey is at best a temporary solution.",

  # CP RECURSION MISCELLANEOUS
  "You must remember that you should always run all the basic diagnostics first.",
  "She announced her victory to irritate her detractors.",
  "Paul cherished his worn-out diary, whose pages were heavily marked by stains",
  "The few that could understand the manifesto's complex worldview appreciated it.",

  # CP RECURSION CENTER EMBEDDING ONE LEVEL
  "The mirror the girl punched shattered into a million pieces.",
  "Every desk I put a heavy lamp on eventually cracked.",
  "Pieces of a diary Paul tore to shreds lay on the ground.",
  "The expert every journalist questioned remained completely silent.",

  # CP RECURSION CENTER EMBEDDING TWO LEVELS
  "The expert every investigator we dispatched questioned remained completely silent.",
  "The astronomy the ancients we revere studied was not separate from astrology.",
  "A candidate the committee we put together rejected developed the imposter syndrome.",
  "The worldview that the prose Nietzsche wrote expressed was unprecedented.",
]

task = "Consider the sentence below and complete the following three tasks:\n\
1. Does the sentence in question contain an instance of recursion? If so, \
identify the recursive part and say what kind of recursion it is. Note that \
there are different types of recursion, e.g. adjectival recursion (an \
adjective modifying an adjective-modified noun), Saxon Genitive a.k.a. \
possessive 's recursion (an 's-possessed 's-possessor), prepositional phrase \
recursion (a prepositional phrase with another prepositional phrase inside), \
clausal recursion (a clause within another clause), and so on.\n\
2. Provide code for a syntactic tree representing the structure of the \
sentence that can be rendered with LaTeX's forest package. Assume X-bar \
theory.\n\
3. If you identified that the sentence contains a recursive structure, \
expand it by adding another layer of recursion (of the same type) to the \
identified structure.\n"

def write_file(iteration_i):
  iteration = iteration_i + 1
  iter_string = "0"
  if iteration < 10:
    iter_string = "0" + str(iteration)
  else:
    iter_string = str(iteration)
  file_name = "/content/drive/MyDrive/Google Drive.lnk/Academic work/" + \
    "AI linguistics/Large linguistic models/02_recursion/rec_prompts/" + \
    "rec" + "_" + iter_string + ".txt"
  print(file_name)
  prompt = task + '"'+recs[iteration_i]+'"'
  print(prompt)
  f = open(file_name, "w")
  f.write(prompt)
  f.close()

for i in range(len(recs)):
  write_file(i)
```
