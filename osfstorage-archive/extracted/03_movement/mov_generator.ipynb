```python
movs = [
  # A-MOVEMENT: PASSIVE VOICE
  "The fence had been painted.",
  "A tree is being observed.",

  # POLAR AUX-QUESTION
  "Have you thought of using a second monitor?",
  "Can I actually survive this?",

  # CONDITIONAL AUX-INVERSTION
  "Were I good at coding, this would be much easier.",
  "They would have won had they trained harder.",

  # NEGATIVE/ONLY AUX-INVERSION
  "Never have I been so excited!",
  "Only now do I recognize him.",

  # WH-QUESTION
  "Who did they invite here?",
  "Why did they let this weirdo enroll?",

  # WH-QUESTION WITH PIED-PIPING
  "What color is the roof of his house?",
  "In what way is he a better candidate?",

  # WH-QUESTION WITH P-STRANDING
  "Who did she vote for in the last election?",
  "What do they believe in?",

  # WH-QUESTION PAST C-BOUNDARY
  "When do you think she will arrive?",
  "What did he say he was going to prepare?",

  # TOPICALIZATION Ā-FRONTING
  "The red book, I have studied assiduously.",
  "For his teachings, I have a lot of respect.",

  # LOCATIVE FRONTING
  "Right behind me stood the Archangel Gabriel.",
  "Under the bed lay a little black kitten.",

  # RAISING ADJECTIVE
  "Natasha was likely to win the election.",
  "The hurricane is certain to wreak havoc.",

  # RAISING TO SUBJECT
  "Mark seemed to be completely losing it.",
  "Luna tends to disagree with me.",

  # THOUGH-MOVEMENT
  "Not all life circumstances are easy to alter.",
  "Radu Jude's latest film was a delight to watch.",

  # RAISING TO OBJECT
  "Susan believes her niece to be the right pick.",
  "I expected the train to be on time.",

  # MOVEMENT IN RELATIVE CLAUSES
  "The child who I gave a lollipop is smiling.",
  "She knows the rough area where he went."
]

task = "Consider the sentence below. Provide code for a syntactic tree \
representing its structure that can be rendered with LaTeX's forest package. \
Assume X-bar theory. Pay special attention to syntactic movement and represent \
all instances of movement with (indexed) traces. Do not write code to draw the \
actual syntactic arrows that connect the traces.\n"

def write_file(iteration_i):
  iteration = iteration_i + 1
  iter_string = "0"
  if iteration < 10:
    iter_string = "0" + str(iteration)
  else:
    iter_string = str(iteration)
  file_name = "/content/drive/MyDrive/Google Drive.lnk/Academic work/" + \
    "AI linguistics/Large linguistic models/03_movement/mov_prompts/" + \
    "mov" + "_" + iter_string + ".txt"
  print(file_name)
  prompt = task + '"'+movs[iteration_i]+'"'
  print(prompt)
  f = open(file_name, "w")
  f.write(prompt)
  f.close()

for i in range(len(movs)):
  write_file(i)
```
